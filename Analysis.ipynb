```python
df["mpC"]=pd.to_numeric(df["mpC"],errors='coerce')
df=df.dropna(subset=["mpC"])
```

```python
print(df["mpC"].head())
```

```python
print("Avearage:",df["mpC"].mean())
print("Max:",df["mpC"].max())
print("Min:",df["mpC"].min())
```

```python
import matplotlib.pyplot as plt
```

```python

plt.hist(df["mpC"])
plt.xlabel("Melting Point (°C)")
plt.ylabel("Frequency")
plt.show()
```

```python
df=df[["name","mpC"]]
```

```python
print("Highest melting point:")
print(df.loc[df["mpC"].idxmax()])

print("\nLowest melting point:")
print(df.loc[df["mpC"].idxmin()])
```

```python
df=df[df["mpC"]<500]  #removes extremly high value
```

```python
plt.hist(df["mpC"],bins=30)
plt.xlabel("Meltting point  (°C)")
plt.ylabel("Frequency")
plt.title("Distribution of Melting points")
plt.show()
```
